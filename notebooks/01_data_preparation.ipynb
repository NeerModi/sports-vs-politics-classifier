# Sports vs Politics Text Classification - Data Preparation

This notebook handles the initial data loading, exploration, and preparation for building a machine learning classifier to distinguish between sports and politics news articles.

## Section 1: Import Libraries and Load Dataset

```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path

# Set style for better visualizations
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

print("Libraries imported successfully!")
```

```python
# Load the JSON dataset
json_file_path = '../data/raw/News_Category_Dataset_v3.json'

# Load JSON file
data = []
with open(json_file_path, 'r') as f:
    for line in f:
        data.append(json.loads(line))

# Create DataFrame
df = pd.DataFrame(data)
print(f"Dataset loaded successfully!")
print(f"Total records: {len(df)}")
print(f"\nDataset shape: {df.shape}")
print(f"\nColumn names: {df.columns.tolist()}")
print(f"\nFirst few rows:")
print(df.head())
```

## Section 2: Explore Dataset and Check Categories

```python
# Check unique categories
print(f"Unique categories in dataset:")
print(f"\n{df['category'].value_counts()}")

# Check for SPORTS and POLITICS categories
sports_count = len(df[df['category'] == 'SPORTS'])
politics_count = len(df[df['category'] == 'POLITICS'])

print(f"\n\nSPORTS articles: {sports_count}")
print(f"POLITICS articles: {politics_count}")
print(f"Total (SPORTS + POLITICS): {sports_count + politics_count}")
```

## Section 3: Filter Data for SPORTS and POLITICS

```python
# Filter only SPORTS and POLITICS categories
df_filtered = df[df['category'].isin(['SPORTS', 'POLITICS'])].copy()

print(f"Filtered dataset shape: {df_filtered.shape}")
print(f"\nClass distribution after filtering:")
print(df_filtered['category'].value_counts())

# Visualize class distribution
fig, ax = plt.subplots(figsize=(10, 6))
df_filtered['category'].value_counts().plot(kind='bar', ax=ax, color=['#FF6B6B', '#4ECDC4'])
ax.set_title('Class Distribution: SPORTS vs POLITICS', fontsize=14, fontweight='bold')
ax.set_xlabel('Category', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
plt.tight_layout()
plt.savefig('../results/class_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

print("\nClass distribution chart saved to results/class_distribution.png")
```

## Section 4: Combine Headline and Description, Encode Labels

```python
# Check columns available
print("Available columns:")
print(df_filtered.columns.tolist())
print("\nSample data:")
print(df_filtered[['category', 'headline', 'short_description']].head(3))
```

```python
# Combine headline and description
df_filtered['text'] = df_filtered['headline'].fillna('') + ' ' + df_filtered['short_description'].fillna('')

# Encode labels: SPORTS = 0, POLITICS = 1
label_mapping = {'SPORTS': 0, 'POLITICS': 1}
df_filtered['label'] = df_filtered['category'].map(label_mapping)

# Create final dataframe with text and label
df_processed = df_filtered[['text', 'label', 'category']].copy()

print(f"Processed dataset shape: {df_processed.shape}")
print(f"\nLabel encoding:")
print(f"SPORTS: 0")
print(f"POLITICS: 1")
print(f"\nLabel distribution:")
print(df_processed['label'].value_counts())

# Show sample data
print(f"\nSample data (first 2 rows):")
for i in range(2):
    print(f"\n--- Row {i+1} ---")
    print(f"Category: {df_processed.iloc[i]['category']}")
    print(f"Label: {df_processed.iloc[i]['label']}")
    print(f"Text: {df_processed.iloc[i]['text'][:200]}...")  # Show first 200 chars
```

## Section 5: Save Processed Data

```python
# Create output directory if it doesn't exist
output_dir = Path('../data/processed')
output_dir.mkdir(parents=True, exist_ok=True)

# Save processed data to CSV
output_file = output_dir / 'sports_politics.csv'
df_processed.to_csv(output_file, index=False)

print(f"✓ Processed data saved to: {output_file}")
print(f"✓ Total records saved: {len(df_processed)}")
print(f"✓ File size: {output_file.stat().st_size / 1024 / 1024:.2f} MB")

# Verify the saved file
df_verify = pd.read_csv(output_file)
print(f"\n✓ Verification - Loaded {len(df_verify)} records from saved file")
print(f"✓ Columns: {df_verify.columns.tolist()}")
print(f"\n✓ Data preparation complete!")
```
